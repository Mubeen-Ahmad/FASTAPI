```python
from fastapi import FastAPI
import uvicorn
import nest_asyncio
import requests


app = FastAPI() # call it


@app.get("/") # decoractor   get request on /\

async def root():
    return {"message": "Hello World"}

@app.get('/index')
async def index():
    return {"message": "index page"}


if __name__ == "__main__":
    nest_asyncio.apply()
    uvicorn.run(app)
```
